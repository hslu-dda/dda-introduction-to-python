## Ein Jahr herunterladen

```python
# Wir schauen uns die Daten von 2024 an.

# url definieren
url = 'https://data.stadt-zuerich.ch/api/3/action/datastore_search?resource_id=6e1405b8-5fa1-4e2f-b9fa-52f0796a35b0&limit=100000'

# mit requests Daten abrufen
response = requests.get(url)
# json-objekt aus der Antwort
data = response.json()

# Dataframe aus dem Json-Objekt
df = pd.DataFrame(data["result"]["records"])
```

```python


# leere Liste
df_ls = [] 

# Für jedes Element in der ID-Liste die Daten herunterladen und in einer DF-Liste speichern
for e,id in enumerate(resource_list): # enumerate() erlaubt, den index in der Liste jeweils anzuzeigen - hilft beim debuggen
    print(e,id) # ausprinten wo in der Liste sind
    url = 'https://data.stadt-zuerich.ch/api/3/action/datastore_search?resource_id='+ id +'&limit=100000' # <- id ist was sich jeweils in der URL ändert

    response = requests.get(url)
    data = response.json()
    this_df = pd.DataFrame(data["result"]["records"])
    df_ls.append(this_df) # <- den DF zur vormals leeren Liste hinzufügen 

```

```python
# Dfs aneinanderkleben
df = pd.concat(df_ls)
# Indexe der einzelnen dfs zusammenfügen
df.reset_index(drop=True)
```

```python
df.to_csv('data/weather/zurich_weather_1992-2026_all_values.csv', index=False)
```
