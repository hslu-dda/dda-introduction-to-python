# Cities Exploration

````python
import pandas as pd
import altair as alt
````

````python
df_full =pd.read_csv('data/cities/merged_wide_complete.csv')
````

````python
df_full.head()
````

````python
# wir erstellen eine Filterliste. Erstes Objekt: die Städtenamen. Zweites: Einwohnerzahl. 
lst = ['cities', 'cpop1_Population on the 1st of January, total']

for col in df_full.columns:
    # filtern nach ctran und zur liste hinzufügen
    if 'ctran' in col:
        lst.append(col)
````

````python
#Spalten anhand der Filterliste anzeigen. in neuem df abspeichern
df = df_full[lst]
df
````

````python
df.columns
````

### Quizz

    1. Balkenchart - Autostadt
        a) sortiere den df nach den meisten einwohner  & zeige die ersten 10 an
        b) speichere das Ergebnis in einem df ```df_autofrei```
        c) erstelle einen Balkenchart mit der anzahl Autos pro 1000 Einwohner auf der x achse und den Städten auf der y-Achse
        d) passe die Properies an so dass man etwas sieht
        e) sortiere die Balken aufsteigend (grösste zuoberst)
        f) setze einen aussagekräftigen Titel
        g) setze ein Tootip

````python
df_autofrei = df.sort_values('cpop1_Population on the 1st of January, total', ascending=False).head(10)
````

````python
alt.Chart(df_autofrei).mark_bar().encode(
    x = 'ctran_Number of registered cars per 1000 population', 
    y = alt.Y('cities', sort='-x'), 
    tooltip=['cities', 'ctran_Number of registered cars per 1000 population']
).properties(
    width = 800, 
    height = 500, 
    title = "Anzahl Autos pro 1000 Einwohner in den 10 grössten Städten im Datensatz"
)
````

2. Scatterplot - Unfälle

        a) erstelle einen Scatterplot mit Anzahl Autos pro 1000 Einwohner vs nutzung des öffentlichen Verkehrs in Prozent
        b) wie oben, passe die Grösse, Titel und Tooltip an
        c) mache die Punkte grösser z.B [```mark_circle(size=300)```]
        d) ändere die Farbe der Punkte nach nutzung des öffentlichen Verkehrs und setze die Farbskala "greenblue"


````python
alt.Chart(df).mark_circle(size=300, 
                        ).encode(
    x = 'ctran_Number of registered cars per 1000 population', 
    y = 'ctran_Share of journeys to work by public transport (rail, metro, bus, tram) -%', 
    tooltip=['cities', 'ctran_Number of registered cars per 1000 population','ctran_Share of journeys to work by public transport (rail, metro, bus, tram) -%' ], 
    color = alt.Color('ctran_Share of journeys to work by public transport (rail, metro, bus, tram) -%:Q').scale(scheme ='greenblue')
).properties(
    width = 500, 
    height = 500, 
    title = "Anzahl Autos pro 1000 Einwohner vs Nutzung Öffentlicher Verkehr"
)
#
````
