## Basic Statistics

Pandas bietet eine ganze Reihe von eingebauten Funktionen um einfache und komplexe Statistiken zu einem Datensatz auszugeben, darunter `.mean()`, `.median()`, `,mode()`, `.min()` und `.max()`.

- Der Mittelwert (Mean) ist der Durchschnitt, der berechnet wird, indem alle Werte addiert und durch die Anzahl der Werte geteilt werden. 
- Der Median hingegen ist der mittlere Wert in einer sortierten Datenreihe, der die Daten in zwei Hälften teilt – unabhängig von extremen Ausreißern. Sollte es sich um eine gerade Anzahl von Werten handeln so ist der Median das Mittel der beiden mittleren Werte. 
- Mode bezeichnet den häufigsten Wert in einer Zahlenreihe. Es kann auch mehrere Modalwerte geben wenn mehrere Zahlen gleich häufig vorkommen. 
- Min. und Max. bezeichnen den niedrigsten, bzw. höchsten Wert in einer sortierten oder unsortierten Zahlenreihe.

### Mean, Median, Min. und Max.

````python
# Calculate the mean
mean_arbeitslosigkeit = df_raw['clma_Unemployment rate'].mean()

# Calculate the median
median_arbeitslosigkeit = df_raw['clma_Unemployment rate'].median()

# Get the min
min_arbeitslosigkeit = df_raw['clma_Unemployment rate'].min()

# Get the max
max_arbeitslosigkeit = df_raw['clma_Unemployment rate'].max()

# Print the results
print(f"Mean Arbeitslosigkeit(%): {mean_arbeitslosigkeit}")
print(f"Median Arbeitslosigkeit(%): {median_arbeitslosigkeit}")
print(f"Min. Arbeitslosigkeit(%): {min_arbeitslosigkeit}")
print(f"Max. Arbeitslosigkeit(%): {max_arbeitslosigkeit}")
````

````python
# Get index of min and max
idx_min = df_raw['clma_Unemployment rate'].idxmin()
idx_max = df_raw['clma_Unemployment rate'].idxmax()

# Get full rows
row_min = df_raw.loc[idx_min]
row_max = df_raw.loc[idx_max]

# Do something with it
print("Row with min:", row_min["cities"])
print("Row with max:", row_max["cities"])
````

### Filter Cheatsheet

#### Comparison Operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | Equal to | `df[df['col'] == 5]` |
| `!=` | Not equal to | `df[df['col'] != 'A']` |
| `>` | Greater than | `df[df['price'] > 100]` |
| `<` | Less than | `df[df['age'] < 30]` |
| `>=` | Greater than or equal | `df[df['score'] >= 90]` |
| `<=` | Less than or equal | `df[df['temp'] <= 0]` |

#### Logical Operators (combining conditions)

| Operator | Meaning | Example |
|----------|---------|---------|
| `&` | AND | `df[(df['a'] > 1) & (df['b'] < 5)]` |
| `\|` | OR | `df[(df['x'] == 'A') \| (df['y'] == 'B')]` |
| `~` | NOT | `df[~(df['status'] == 'inactive')]` |

#### Membership Operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `.isin()` | Value in list | `df[df['col'].isin(['A', 'B', 'C'])]` |
| `.between()` | In range | `df[df['age'].between(18, 35)]` |

#### String Operators (via `.str` accessor)

```python
df[df['name'].str.contains('test')]    # substring match
df[df['name'].str.startswith('A')]     # starts with
df[df['name'].str.endswith('z')]       # ends with

Key Notes
- Each comparison produces a boolean Series (True/False per row).
- Use parentheses around each condition when combining with & or |.
- Use & and |, not and/or (those don't work element-wise).
- ~ is the bitwise NOT — use it instead of not for negation.

## Missing Data

Often data is sparse so understanding were the data coverage is better/worse helps to decide on what aspect to focus on. 

````python
# radical approach, drop all rows/columns with missing data
only_complete = df_raw.dropna()
only_complete.head()
````

Add a threshold of how many at least must be filled. 
These are the cities were from 149 possible values, 2/3 are available, not so bad..



````python
print("columns: ", len(df_raw.columns))
atleast = df_raw.dropna(thresh=100)

atleast.head()
````

A more sophisticated approach calculates which cities or indicators have the best coverage. This can be done in three steps: 

- `notna()` creates a boolean DataFrame: True where a cell has a value, False where it's missing.
- `.sum(axis=1)` sums across each row or column (axis=1 = rows, axis=0 = columns), counting how many True values each city or indicator has.
- `.sort_values(ascending=False)` — sorts from most to fewest filled.
- because the df returns the first row, which is the index row, the city names are no longer visible. This can be changed using `.set_index("cities")`

````python
print("row count:", len(df_raw))
indicator_coverage = df_raw.notna().sum(axis=0).sort_values(ascending=False)
indicator_coverage.head()
````

````python
print("column count:", len(df_raw.columns))
city_coverage = df_raw.set_index("cities").notna().sum(axis=1).sort_values(ascending=False)
city_coverage.head()
````
