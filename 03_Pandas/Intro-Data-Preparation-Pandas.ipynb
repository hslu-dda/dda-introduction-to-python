# Introduction to Data Preparation with Pandas

Im Rahmen von Think 3 bereiten wir die Daten so auf, dass sie sich für die Visuelle Analyse eignen. D.h. wir rechnen Werte teils so um, dass sie sich für die Visualisierung eignen. 

## Setup

```python
import pandas as pd
```

```python
df_raw = pd.read_csv("data/eurostat_european_cities.csv")
df_raw.head()
```

## Subset Erstellen

```python
# erstelle eine Kopie um auf das Original zurückgreifen zu können
df = df_raw.copy()
city_selection = [ "Berlin", "Paris (greater city)", "Budapest (greater city)", "Ljubljana", "Bruxelles/Brussel (greater city)"]

df = df[df["cities"].isin(city_selection)]
df.head()
```

```python
col_selection = [
    "cities",
    "cpop1_Population on the 1st of January, total",
    "clma_Persons unemployed, total",
    "cpopcb_Nationals",
    "cpopcb_EU foreigners",
    "cpopcb_Non-EU foreigners"
    ]

df = df[col_selection]
df.head()
```

## Spalten ersetzen oder hinzufügen

Um eine neue Spalte hinzu zufügen kann `insert()` genutzt werden wo drei Argumente mitgegeben werden müssen: Position, Name und Inhalt der Spalte.


```python
df_ex1 = df.copy()

df_ex1.insert(
    df_ex1.columns.get_loc("cpopcb_Nationals"), # index position
    "cpopcb_Nationals scaled by 10'000",
    (df_ex1["cpopcb_Nationals"] / 10000).round(0)
)

df_ex1.head()
```

Um eine Spalte mit neuen Werten zu füllen und ihren namen zu ändern kann `rename()` verwendet werden. 

```python
df_ex2 = df.copy()

# Werte überschreiben
df_ex2["cpopcb_Nationals"] = (df_ex2["cpopcb_Nationals"]/10000).round(0)

# Umbenennen
df_ex2.rename(columns={"cpopcb_Nationals":"cpopcb_Nationals scaled by 10'000"}, inplace=True)

df_ex2.head()
```

## Datenaufbereitung & Werte berechnen

### Summe und Mittelwerte hinzufügen

Mittels `.loc[]` können zusätzliche Zeilen hinzugefügt werden deren Index anstelle einer Nummer ein Name ist (bspw. "sum"). Das Selbe kann auch für Durschnittswerte gemacht werden. Damit die Werte dieser neuen Zeilen einander nicht beeinflussen müssen sie getrennt berechnet werden. 

```python
df_sum = df.copy()
df_avg = df.copy()

df_sum.loc["sum"] = df_sum.sum(numeric_only=True)
df_sum.loc["avg"] = df_avg.mean(numeric_only=True).round(2)

df_sum.head(10)
```

### Werte Skalieren

Für die Visualisierung ist es manchmal hilfreich wenn die Werte um einen bestimmten Faktor skaliert sind, so dass die Verhältnisse besser in visuelle Mittel wir Grösse, Position oder Farbe übersetzbar sind. 

Dazu kann `insert()` genutzt werden wo drei Argumente mitgegeben werden müssen: Position, Name und Inhalt der Spalte.

```python
df_scaled = df.copy()

# Population
df_scaled["cpop1_Population on the 1st of January, total"] = (df_scaled["cpop1_Population on the 1st of January, total"]/10000).round(0)
df_scaled.rename(columns={"cpop1_Population on the 1st of January, total":"cpop1_Population on the 1st of January, total scaled by 10'000"}, inplace=True)

# Employment
df_scaled["clma_Persons unemployed, total"] = (df_scaled["clma_Persons unemployed, total"]/10000).round(0)
df_scaled.rename(columns={"clma_Persons unemployed, total":"clma_Persons unemployed, total scaled by 10'000"}, inplace=True)

# Nationals
df_scaled["cpopcb_Nationals"] = (df_scaled["cpopcb_Nationals"]/10000).round(0)
df_scaled.rename(columns={"cpopcb_Nationals":"cpopcb_Nationals, total scaled by 10'000"}, inplace=True)

# Übung: Wiederhole das ganze für cpopcb_EU foreigners und cpopcb_Non-EU foreigners

# Foreigners
df_scaled["cpopcb_EU foreigners"] = (df_scaled["cpopcb_EU foreigners"]/10000).round(0)
df_scaled.rename(columns={"cpopcb_EU foreigners":"cpopcb_EU foreigners, total scaled by 10'000"}, inplace=True)

# Non-EU Foreigners
df_scaled["cpopcb_Non-EU foreigners"] = (df_scaled["cpopcb_Non-EU foreigners"]/10000).round(0)
df_scaled.rename(columns={"cpopcb_Non-EU foreigners":"cpopcb_Non-EU foreigners, total scaled by 10'000"}, inplace=True)

df_scaled.head()
```

### Prozente Berechnen

population_share = (population / sum(all population)) * 100

```python
df_perc = df.copy()

# Population
df_perc["cpop1_Population on the 1st of January, total"] = (
    df_perc["cpop1_Population on the 1st of January, total"] /
    df_perc["cpop1_Population on the 1st of January, total"].sum()*100
    ).round(0)

df_perc.rename(columns={"cpop1_Population on the 1st of January, total":
    "cpop1_Population on the 1st of January, total in %"}, inplace=True)


# Unemployment
df_perc["clma_Persons unemployed, total"] = (
    df_perc["clma_Persons unemployed, total"] /
    df_perc["clma_Persons unemployed, total"].sum()*100
    ).round(0)

df_perc.rename(columns={"clma_Persons unemployed, total":
    "clma_Persons unemployed, total in %"}, inplace=True)

# Nationals
df_perc["cpopcb_Nationals"] = (
    df_perc["cpopcb_Nationals"] /
    df_perc["cpopcb_Nationals"].sum()*100
    ).round(0)

df_perc.rename(columns={"cpopcb_Nationals":
    "cpopcb_Nationals in %"}, inplace=True)

# EU Foreigners
df_perc["cpopcb_EU foreigners"] = (
    df_perc["cpopcb_EU foreigners"] /
    df_perc["cpopcb_EU foreigners"].sum()*100
    ).round(0)

df_perc.rename(columns={"cpopcb_EU foreigners":
    "cpopcb_EU foreigners in %"}, inplace=True)

# Non-EU Foreigners
df_perc["cpopcb_Non-EU foreigners"] = (
    df_perc["cpopcb_Non-EU foreigners"] /
    df_perc["cpopcb_Non-EU foreigners"].sum()*100
    ).round(0)

df_perc.rename(columns={"cpopcb_Non-EU foreigners":
    "cpopcb_Non-EU foreigners in %"}, inplace=True)


df_perc.head(10)
```

### Abweichung zum Mittelwert berechnen

```python
# Wir verwenden hierfür die Prozente anstelle der absoluten Zahlen
df_diff = df_perc.copy()

# using a for loop for faster calculations
for col in df_diff.columns:
    
    if col != 'cities':
        # werte berechnen
        df_diff[col] = (df_diff[col] - df_diff[col].mean()).round(2)
        # umbenennen
        new_name = col + ' Difference %'
        df_diff.rename(columns={col : new_name}, inplace=True)
        
df_diff.head(10)
```

### Kategorien bilden

Durch die Anwendung von Schwellenwerten oder Bereichen, die auf dem Wertebereich der jeweiligen Spalte basieren, können Kategorien gebildet werden. Das Ziel ist es, aus den kontinuierlichen Werten diskrete Kategorien zu schaffen, die für die Visualisierung und die spätere Arbeit mit MID besser handhabbar sind. 

Wir können dazu die Funktion `cut` aus [Pandas nutzen](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.cut.html). 

```python
df_cat = df.copy()

# create three bins of equal size
bins = 3
labels = ["small", "medium", "large"]

for col in df_cat.columns:
    if col != 'cities':
        # werte berechnen
        df_cat[col] = pd.cut(df_cat[col], bins=bins, labels=labels)
        # umbenennen
        new_name = col + ' cat'
        df_cat.rename(columns={col : new_name}, inplace=True)
        
df_cat.head(10)
```
