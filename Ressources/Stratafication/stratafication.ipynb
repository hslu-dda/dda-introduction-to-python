# Stratified Sampling

Stratified Sampling is a sampling technique used to obtain samples that best represent the population. It reduces bias in selecting samples by dividing the population into homogeneous subgroups called strata, and randomly sampling data from each stratum(singular form of strata). In statistics, stratified sampling is used when the mean values of each stratum will differ.

For example, the VOTO dataset contains answers from people who were asked questions after voting. We can use strafification to create a subsample were, for example, the age groups are representative for the age distribution across the whole dataset. 

- **Goal:** We need to get a representative sample with a size of 1500 data points of the data distribution for the properties `age_category`, `part` and `educ`.

For more infos see [here](https://codepointtech.com/mastering-stratified-sampling-in-pandas-for-representative-data/) and [here](https://www.geeksforgeeks.org/python/stratified-sampling-in-pandas/)

```python
import pandas as pd
# avoid scientific notation number format
pd.set_option("display.float_format", lambda x: f"{x:.6f}")
```

```python
#loading data
voto_project = "./data/data_reduced_coding1.csv"
df = pd.read_csv(voto_project)

print("data set length:", (len(df)))
df.head()
```

## Understanding data distribution

We can use the `value_counts` function to understand the distribution of each category in our dataset. This will help us validate the subsample we created afterwards.

```python
print("\nOriginal age_category Distribution:")
print(df["age_category"].value_counts(normalize=True))
print("\nOriginal part Distribution:")
print(df["part"].value_counts(normalize=True))
print("\nOriginal educ Distribution:")
print(df["educ"].value_counts(normalize=True))
```

## Proportionate sampling

Using pandas groupby, separate the participants into groups based on their age_category and random sample from each group based on population proportion. The total sample size is 60%(0.6) of the data set size

```python
prop_sample = df.groupby("age_category").apply(lambda x: x.sample(frac=0.6, replace=False))
prop_sample.head()
```

Now compare the samples data distribution with the above distribution to validate our sampling process. It is surprisingly accurate! 🤗

```python
print("sample data length", len(prop_sample))
print("\nSample age category Distribution:")
print(prop_sample["age_category"].value_counts(normalize=True))
```

## Proportionate sampling for multiple values

What if you need to stratify by more than one characteristic? For example, you might want to ensure representation by "educ" AND "age_category" simultaneously. Pandas handles this elegantly by passing a list of column names to groupby().

```python
prop_sample_2 = df.groupby(["age_category", "part", "educ"]).apply(lambda x: x.sample(frac=0.6, replace=False))
prop_sample_2.head()
```

again we can verify the distributions of our samples data with the complete data distribution. Yaay! 🎈

```python
print("\n Sample age_category Distribution:")
print(prop_sample_2["age_category"].value_counts(normalize=True))
print("\nSample part Distribution:")
print(prop_sample_2["part"].value_counts(normalize=True))
print("\n Sample educ Distribution:")
print(prop_sample_2["educ"].value_counts(normalize=True))
```

## Set sample size & export

We use the `frac` value to approach a specific sample size, e.g. 1500

```python
final_sample = df.groupby(["age_category", "part", "educ"], group_keys=False).apply(lambda x: x.sample(frac=0.03194, replace=False))
print(len(final_sample))

final_sample
```

```python
final_sample.to_csv("./export/proportional-sample.csv", index=False)
```
