## NLP Techniques

### Tokenization

Tokenization is the task of splitting a text into meaningful segments, called tokens. The algorithm uses these as "tokens" of analysis.

```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for token in doc:
  print(token)
```

### Sentence Segmentation

With Spacy it is also possible to split the text into sentences. Full documentation can be seen [here](https://spacy.io/usage/linguistic-features#sbd).

```python
doc = nlp("This is a sentence. This is another sentence.")
assert doc.has_annotation("SENT_START") # is a safety check to ensure that the Doc object (doc) has sentence boundary annotations (SENT_START) before attempting to iterate over the sentences in doc.sents.
for sent in doc.sents:
    print(sent.text)
```

### Part of Speech Tagging (POS)

- `.lemma` base or dictionary form of the token (e.g. "run" for "running")
- `.pos` high level category such as noun, verb, adjective etc. See [here](https://universaldependencies.org/u/pos/).
- `.shape` The shape of the token, which abstracts its capitalization and punctuation patterns. E.g. "Xxxxx" (for "Apple"), "xxx" (for "run").
- `.is_alpha` A boolean indicating whether the token consists only of alphabetic characters. E.g. True (for "running"), False (for "123" or ".")
- `.is_stop` A boolean indicating whether the token is a stop word. Stop words are common words like “the”, “is”, “and” that are often ignored in NLP tasks. E.g. True (for "is"), False (for "running").

See the full documentation [here](https://spacy.io/usage/linguistic-features#pos-tagging).

```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for token in doc:
    print(token.text, token.lemma_, token.pos_,
          token.shape_, token.is_alpha, token.is_stop)
```

### Named Entity Recognition (NER)

- `CARDINAL`: Numerals that do not fall under another type.
- `DATE`: Absolute or relative dates or periods.
- `EVENT`: Named hurricanes, battles, wars, sports events, etc.
- `FAC`: Buildings, airports, highways, bridges, etc.
- `GPE`: Countries, cities, states.
- `LANGUAGE`: Any named language.
- `LAW`: Named documents made into laws.
- `LOC`: Non-GPE locations, mountain ranges, bodies of water.
- `MONEY`: Monetary values, including unit.
- `NORP`: Nationalities or religious or political groups.
- `ORDINAL`: "First", "second", etc.
- `ORG`: Companies, agencies, institutions, etc.
- `PERCENT`: Percentage, including "%".
- `PERSON`: People, including fictional.
- `PRODUCT`: Objects, vehicles, foods, etc. (not services).
- `QUANTITY`: Measurements, as of weight or distance.
- `TIME`: Times smaller than a day.
- `WORK_OF_ART`: Titles of books, songs, etc.

Full documentation can be seen [here](https://spacy.io/usage/linguistic-features#named-entities).

```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

Spacy comes with some visualization options to better understand the text structure. All options can be seen in the documentation [here](https://spacy.io/usage/visualizers/).

```python
displacy.render(doc, style="ent", jupyter=True)
```

```python
# 2. load the machine learning model and save it into a variable called `nlp_de`
nlp_de = spacy.load("de_core_news_md")
```

```python
# 3. load the dataset into a dataframe
df = pd.read_csv("./data/parl_speech_V2_sample.csv")
df.head()
```

```python
# get one row from the dataframe
doc = nlp_de(df["text"][0])

# use the visualization for some overview
displacy.render(doc, style="ent", jupyter=True)
```

## Save the NER results as JSON in a CSV

```python
import json 
# get the entities from the document
entities = []

# loop over the entities
for ent in doc.ents:
    # test if you already saved the same entity (avoid duplicates)
    if ent.text not in entities:
        # create a object for each entity
        obj = {
            "entity": ent.text,
            "label": ent.label_
        }
        # save it into an array/list
        entities.append(obj)
        
# output the entities in json format
print(json.dumps(entities))
```

```python
# add to the df
df.at[0, "entities"] = json.dumps(entities)
df.to_csv("./parl_speech_V2_sample_entities.csv", index=False)
```
