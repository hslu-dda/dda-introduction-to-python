```python
#import libraries
import altair as alt
import pandas as pd
```

```python
#load data
df = pd.read_csv('data/Algiers_Weather.csv')
```

### Data Cleaning

```python
#check colum data types
df.dtypes
```

```python
# format as date
df["time"] = pd.to_datetime(df["time"], format="%d.%m.%Y")

# rename time to date
df = df.rename(columns= {'time':'date'} )
```

```python
df.dtypes
```

```python
df
```

```python
# wir schauen uns die Werte ab 2012 an
df_10years = df[df['date']>'2011-12-31']
df_10years
```

## Charts anordnen & überlagern

```python
temp_mean = alt.Chart(df_10years).mark_line(opacity=0.5, color='green').encode(
    x = 'date:T', 
    y = 'temperature_2m_mean (°C)'
).properties(
    title = 'Mean Tempeature', 
    width = 200, 
    height = 200
)

temp_mean 
```

```python
temp_min = alt.Chart(df_10years).mark_line(opacity=0.5, color='blue').encode(
    x = 'date:T', 
    y = 'temperature_2m_min (°C)'
).properties(
    title = 'Min Tempeature',
    width = 200, 
    height = 200
)
temp_min
```

```python
temp_max = alt.Chart(df_10years).mark_line(opacity=0.5, color='red').encode(
    x = 'date:T', 
    y = 'temperature_2m_max (°C)'
).properties(
    title = 'Max Tempeature',
    width = 200, 
    height = 200
)
temp_max
```

Anordnen von Charts
* `|` nebeneinander
* `&` untereinander
* `+` Überlagerung

```python
temp_min | temp_mean | temp_max
```

```python
temp_min & temp_mean | temp_max
```

```python
# übereinander

temp_min  + temp_max # + temp_mean
```

## Interaktion
### Range Selection
Wir erstellen 2 charts, mit dem unteren können wir auswählen welcher Bereich im Oberen angezeigt wird. 

```python
# datumsreihe formatieren
date_range = (df_10years['date'].min(), df_10years['date'].max())

# Auswahl definieren
brush = alt.selection_interval(
    encodings = ['x'], 
    value={'x':date_range}
    )

# Basis Chart
base = alt.Chart(df_10years).mark_area(opacity=0.5, color='green').encode(
    x = 'date:T', 
    y = 'temperature_2m_mean (°C)'
).properties(
    width=600, 
    height=300
)
# Selektion Anzeigen
upper = base.encode(
    alt.X('date:T').scale(domain=brush)
).properties(
    title = "Durchschnittstempereatur in Algier, 2012 - 2023"
)
# Ganze Zeitreihe
lower  = base.properties(
    height = 60
).add_params(brush)

# beide Charts ausgeben
upper & lower
```
